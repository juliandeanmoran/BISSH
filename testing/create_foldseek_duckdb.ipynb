```python
import duckdb
import os

import numpy as np
import polars as pl

from dotenv import load_dotenv

# Env
load_dotenv("../.env", override=True)
INSTALL_PATH = os.environ["INSTALL_PATH"]
```

```python
# ============================================================
#                            Args
# ============================================================

args = {
    "data_file_fs_clusterSims": f"{INSTALL_PATH}/data_local/6-all-vs-all-similarity-queryId_targetId_eValue.gz",
    "db_path_fs": f"{INSTALL_PATH}/data_local/fs_clusters_db.duckdb",
    "db_path_fs_pq": f"{INSTALL_PATH}/data_local/fs_clusters_db.parquet",
    "name_table_clusterSim": "all_vs_all_similarity"
}
```

```python
# ============================================================
#                     Create FoldSeek DB
# ============================================================

con = duckdb.connect(args["db_path_fs"])
con.execute(f"""
CREATE TABLE IF NOT EXISTS {args[name_table_clusterSim]} AS
SELECT
    column0 AS queryID,
    column1 AS targetID,
    column2 AS eValue
FROM read_csv(
    '{args["data_file_fs_clusterSims"]}',
    delim='\t',
    header=False
)
""")
```

```python
# ============================================================
#                             Test
# ============================================================

threshold = 1_000_000
result = con.execute(f"""
SELECT *
FROM {args["name_table_clusterSim"]}
WHERE eValue >= {threshold}
""").fetchdf()
```
