```python
import pandas as pd
```

```python
df_refseq=pd.read_excel("../data/defense_finder/df-RefSeq.xlsx")
df_refseq
```

```python
df_refseq_na_values = df_refseq[df_refseq['accession_in_sys'].isna()]
df_refseq_na_values
```

```python
# Remove all rows with NA
df_refseq_no_na=df_refseq[df_refseq['accession_in_sys'].notna()][["sys_id","sys_beg","sys_end","type","subtype","accession_in_sys","species"]]
df_refseq_no_na
```

### Clean up

- The "accession_in_sys" column may contain comma-separated lists of accessions in each row.
- Some accessions appear multiple times across rows, possibly with different "type", "subtype", or "species". The sys_id, sys_beg and sys_end will remain unique
- For clean up:
    - Split those comma-separated accessions into separate rows.
    - Deduplicate based on individual accessions.
        - Group by each unique accession and aggregate other columns into lists of ALL values:
        
            `../data/defense_finder/2025-06-20_df_refseq_unique_accessions.tsv`
        - Group by each unique accession and aggregate other columns into lists of ALL values, also count replicate accessions and rank them from most to least prevalent:
        
            `../data/defense_finder/2025-06-24_df_refseq_unique_accessions_ranked.tsv`
        - Group by each unique accession and aggregate other columns into lists of their unique values. Also count replicate accessions and rank them from most to least prevalent:
        
            `../data/defense_finder/2025-06-24_df_refseq_unique_accessions_w_unique_values_ranked.tsv`

```python
# Step 1: Split the 'accession' column into multiple rows
df_refseq_no_na_expanded = df_refseq_no_na.assign(accession_in_sys=df_refseq_no_na['accession_in_sys'].str.split(',')).explode('accession_in_sys')
df_refseq_no_na_expanded
```

```python
# Step 2: Strip whitespace (just in case)
df_refseq_no_na_expanded['accession_in_sys'] = df_refseq_no_na_expanded['accession_in_sys'].str.strip()
```

```python
# Step 3: Get duplicate accession counts
accession_counts = df_refseq_no_na_expanded['accession_in_sys'].value_counts().rename_axis('accession_in_sys').reset_index(name='duplicate_count')
accession_counts
```

```python
# Step 4: Group by accession and aggregate other columns into lists
df_refseq_no_na_grouped = df_refseq_no_na_expanded.groupby('accession_in_sys', as_index=False).agg({
    'type': lambda x: ','.join(x),
    'subtype': lambda x: ','.join(x),
    'species': lambda x: ','.join(x),
    "sys_id": lambda x: ','.join(x),
    "sys_beg": lambda x: ','.join(x),
    "sys_end": lambda x: ','.join(x)
})


# Step 5: Merge the accession counts with the merged dataframe
df_refseq_no_na_grouped_w_counts = df_refseq_no_na_grouped.merge(accession_counts, on='accession_in_sys')

# Step 6: Add the rank and sort the dataframe in descending order
df_refseq_no_na_grouped_w_counts['rank'] = df_refseq_no_na_grouped_w_counts['duplicate_count'].rank(method='dense', ascending=False).astype(int)
df_refseq_no_na_grouped_w_counts = df_refseq_no_na_grouped_w_counts.sort_values(by='rank', ascending=True).reset_index(drop=True)
df_refseq_no_na_grouped_w_counts
```

```python
# Step 7: Repeat the grouping of accession but this time only retain unique type, subtype and species
df_refseq_no_na_grouped_uniq = df_refseq_no_na_expanded.groupby('accession_in_sys', as_index=False).agg({
    'type': lambda x: ','.join(set(x)),
    'subtype': lambda x: ','.join(set(x)),
    'species': lambda x: ','.join(set(x)),
    "sys_id": lambda x: ','.join(set(x)),
    "sys_beg": lambda x: ','.join(set(x)),
    "sys_end": lambda x: ','.join(set(x))
})

# Step 8: Merge the accession counts with the merged dataframe
df_refseq_no_na_grouped_uniq_w_counts = df_refseq_no_na_grouped_uniq.merge(accession_counts, on='accession_in_sys')

# Step 9: Add the rank and sort the dataframe in descending order
df_refseq_no_na_grouped_uniq_w_counts['rank'] = df_refseq_no_na_grouped_uniq_w_counts['duplicate_count'].rank(method='dense', ascending=False).astype(int)
df_refseq_no_na_grouped_uniq_w_counts = df_refseq_no_na_grouped_uniq_w_counts.sort_values(by='rank', ascending=True).reset_index(drop=True)
```

```python
df_refseq_no_na_grouped_uniq_w_counts
```

```python
# Write to csv file
with pd.ExcelWriter('../data/defense_finder/2025-06-24_df_refseq_clean_up.xlsx') as writer:   
    df_refseq_na_values.to_excel(writer,sheet_name="no_accs",index=False)
    df_refseq_no_na_grouped.to_excel(writer,sheet_name="uniq_accs",index=False)
    df_refseq_no_na_grouped_w_counts.to_excel(writer,sheet_name="uniq_accs_ranked",index=False)
    df_refseq_no_na_grouped_uniq_w_counts.to_excel(writer,sheet_name="uniq_accs_w_uniq_values_ranked",index=False)
```
